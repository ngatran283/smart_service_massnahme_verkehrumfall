### Ergebnis der Datenprüfung
Die Prüfung der eindeutigen Werte zeigt folgende Probleme:
- Uneinheitliche fehlende Werte: NaN, Unknown, unknown, na, Unknown or other
- Überflüssige Leerzeichen
- Kodierungsfehler: Lorry (41?100Q)
- Fehlerhafte Werte: „Rural village areasOffice areas" (zusammengefügt), „Tangent road with mountainous terrain and" (abgeschnitten)
- Unklare Codes: 5 und 7 bei Fahrzeugmängeln (Defect_of_vehicle)
- Zielvariable stark unausgeglichen (value_counts): 84.6 % leicht, 14.2 % schwer, 1.3 % tödlich. Da tödliche Unfälle mit 1.3% für Gruppenvergleiche zu selten sind, werden "Serious" und "Fatal" zu schwer zusammengefasst.

## 2 Datenbereinigung
- Leerzeichen am Anfang/Ende entfernt
- Unterschiedliche Schreibweisen fehlender Werte (`unknown`, `na`, `Unknown or other`, leere Zellen) einheitlich als **„Unknown"** kodiert
- Fehlerhafte Zeichen (`?` in Fahrzeugtypen) und fehlerhafte Kategorien (`Rural village areasOffice areas`) korrigiert
- Exakte Duplikate: keine

```python
df_clean = df.copy()

# Spalten mit Text auswählen (außer Uhrzeit)
text_cols = df_clean.select_dtypes(exclude='number').columns.drop('Time')

# 1. Leerzeichen entfernen
for col in text_cols:
    df_clean[col] = df_clean[col].str.strip()

# 2. Fehlende Werte vereinheitlichen
df_clean[text_cols] = df_clean[text_cols].replace(
    ['unknown', 'na', 'Unknown or other'], 'Unknown')
df_clean[text_cols] = df_clean[text_cols].fillna('Unknown')

# 3. Fehlerhafte Werte korrigieren
df_clean['Type_of_vehicle'] = df_clean['Type_of_vehicle'].str.replace('?', '-', regex=False)
df_clean['Area_accident_occured'] = df_clean['Area_accident_occured'].replace(
    'Rural village areasOffice areas', 'Unknown')
df_clean['Lanes_or_Medians'] = df_clean['Lanes_or_Medians'].replace('other', 'Other')
```

## Speichern

```python
df_clean.to_parquet("../../data/df_clean_raw.parquet", index=False)
print("Gespeichert: df_clean_raw.parquet", df_clean.shape)
```
